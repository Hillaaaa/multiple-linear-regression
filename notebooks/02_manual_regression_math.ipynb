```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python

X = np.array([1,2,3])
y = np.array([50, 60, 85])
x_mean = np.mean(X)
y_mean = np.mean(y)
numerator = np.sum((X-x_mean)*(y-y_mean))
denominator = np.sum((X-x_mean)**2)
slope = numerator/denominator
intercept = y_mean -(slope*x_mean)
print(slope)
print(intercept)

model = LinearRegression()
model.fit(X.reshape(-1,1), y)
print(model.coef_)
print(model.intercept_)
```

```python
actuals = np.array([5, 8, 11, 6])
predicted = np.array([6, 7, 12, 4])

residuals = actuals - predicted
mae_calculated = np.mean(np.abs(residuals))
squared_residuals = np.square(residuals)
mse_calculated = np.mean(squared_residuals)
rmse_calculated = np.sqrt(mse_calculated)
squared_distances_from_mean = np.square(actuals-np.mean(actuals))
r2_calculated = 1 - (np.sum(squared_residuals)/np.sum(squared_distances_from_mean))

mae_sklearn = mean_absolute_error(actuals, predicted)
mse_sklearn = mean_squared_error(actuals, predicted)
rmse_sklearn = np.sqrt(mse_sklearn)
r2_sklearn = r2_score(actuals, predicted)

print(f"{"Metric":<15}{"Calculated":<15}{"Sklearn":<15}")
print(f"{"MAE":<15}{mae_calculated:<15.4f}{mae_sklearn:<15.4f}")
print(f"{"MSE":<15}{mse_calculated:<15.4f}{mse_sklearn:<15.4f}")
print(f"{"RMSE":<15}{rmse_calculated:<15.4f}{rmse_sklearn:<15.4f}")
print(f"{"R2":<15}{r2_calculated:<15.4f}{r2_sklearn:<15.4f}")
```
